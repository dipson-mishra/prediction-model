```python
# Encoding categorical features that I will use in model.
```

```python
matches["date"] = pd.to_datetime(matches["date"])
```

```python
matches['venue_code']= matches['venue'].astype('category').cat.codes
```

```python
matches['opp_code']=matches['opponent'].astype('category').cat.codes
```

```python
matches['hour']=matches['time'].str.split(':').str[0].astype(int)
```

```python
matches['day_code']=matches['date'].dt.dayofweek
```

```python
matches
```

```python
matches["target"] = (matches["result"] == "W").astype("int")
```

```python
# Defininf the model and setting parameters. Splitting data into train and test sets.
```

```python
model= RandomForestClassifier(n_estimators=100,min_samples_split=10,random_state=11)
```

```python
train=matches[matches['date']<'2024-01-01']
test=matches[matches['date']>='2024-01-01']
```

```python
# Defininf the features to be used and fitting them in model.
```

```python
features=['venue_code','opp_code','hour','day_code']
```

```python
model.fit(train[features],train['target'])
```

```python
# Making predictions.
```

```python
preds=model.predict(test[features])
```

```python
# Checking accuracy of prediction and making a confusion matrix to see false positives and true negatives.
```

```python
acc=accuracy_score(test['target'],preds)
```

```python
# grouping teams to calculate rolling averages of features to improve prediction.
```

```python
matches_teamwise=matches.groupby('team')
```

```python
group=matches_teamwise.get_group('Barcelona').head(10)
```

```python
def rolling_averages(group,cols,new):
    group=group.sort_values('date')
    roll_stats=group[cols].rolling(3,closed='left').mean()
    group[new]=roll_stats
    group=group.dropna(subset=new)
    return group
```

```python
# defining features that will require rolling averages.
```

```python
cols = ["gf", "ga", "sh", "sot", "dist", "fk", "pk", "pkatt"]
new_cols = [f"{c}_rolling" for c in cols]

rolling_averages(group, cols, new_cols)
```

```python
matches_roll=matches.groupby('team').apply(lambda x: rolling_averages(x, cols, new_cols))
```

```python
matches_roll.droplevel('team').reset_index()
```

```python
# defining the function to predict outcome using rolling averages of features as a feature in model as well.
def predict_outcome(data, predictors):
    model = RandomForestClassifier(n_estimators=100, min_samples_split=10, random_state=11)
    train = data[data["date"] < '2024-01-01']
    test = data[data["date"] >= '2024-01-01']
    model.fit(train[predictors], train["target"])
    preds = model.predict(test[predictors])
    combined = pd.DataFrame(dict(actual=test["target"], predicted=preds), index=test.index)
    error = precision_score(test["target"], preds)
    return combined, error
```

```python
combined,precision=predict_outcome(matches_roll,features + new_cols)
```
